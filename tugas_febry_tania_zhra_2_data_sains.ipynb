## Bagian 4 — Data Dictionary
### 4a. Kerangka otomatis

```python
# TODO: buat kerangka data dictionary otomatis untuk SEMUA kolom:
# variabel, dtype pandas, jumlah nilai unik, jumlah kosong, min, max (khusus numerik), dan contoh nilai.
# Petunjuk: loop atas df.columns; gunakan s.dtype, s.nunique(), s.isna().sum(), pd.api.types.is_numeric_dtype(s), s.unique()[:4]

# Bagian 4a: Kerangka Data Dictionary otomatis

for col in df.columns:
    s = df[col]

    # Mengambil nilai minimum dan maksimum khusus kolom numerik
    if pd.api.types.is_numeric_dtype(s):
        nilai_min = s.min()
        nilai_max = s.max()
    else:
        nilai_min = None
        nilai_max = None

    # Menampilkan informasi setiap kolom
    print("Variabel:", col)
    print("Tipe data:", s.dtype)
    print("Jumlah nilai unik:", s.nunique())
    print("Jumlah kosong:", s.isna().sum())
    print("Nilai minimum:", nilai_min)
    print("Nilai maksimum:", nilai_max)
    print("Contoh nilai:", s.unique()[:4])
    print("-" * 40)
```

### 4c. Dari data dictionary ke aturan validasi
Ubah isian `nilai_valid` Anda menjadi **minimal 3 aturan** yang bisa dijalankan:
- 1 aturan **rentang numerik**,
- 1 aturan **himpunan kategori**,
- 1 aturan **konsistensi antarkolom** (mis. kecepatan implisit untuk moda tertentu).

Gunakan kolom yang **berbeda** dari contoh demo dosen bila memungkinkan.

```python
# TODO: lengkapi kamus aturan. Setiap aturan adalah fungsi yang menerima df dan mengembalikan Series boolean (True = lolos)

# Bagian 4c: Aturan Validasi Data

aturan_saya = {
    # Aturan 1: Jarak ke kampus harus 0 sampai 100 km
    "rentang_jarak": lambda d: (
        d["jarak_km"].notna()
        & d["jarak_km"].between(0, 100)
    ),

    # Aturan 2: Moda transportasi harus termasuk kategori yang dikenal
    "kategori_transportasi": lambda d: (
        d["moda_transportasi"].isin([
            "Berjalan kaki",
            "Sepeda motor",
            "Ojek online"
        ])
    ),

    # Aturan 3: Kecepatan perjalanan tersirat tidak boleh lebih dari 120 km/jam
    "konsistensi_jarak_waktu": lambda d: (
        d["jarak_km"].notna()
        & d["waktu_tempuh_menit"].notna()
        & (d["waktu_tempuh_menit"] > 0)
        & (
            d["jarak_km"] / (d["waktu_tempuh_menit"] / 60)
            <= 120
        )
    )
}

for nama_aturan, fungsi in aturan_saya.items():
    lolos = fungsi(df).fillna(False)
    print(
        f"{nama_aturan}: "
        f"{int((~lolos).sum())} baris melanggar"
    )

    # Menampilkan nomor baris, bukan nama responden
    print("Indeks baris:", df.index[~lolos].tolist())
    print("-" * 40)

```

Berdasarkan hasil validasi, terdapat 4 baris yang melanggar aturan rentang jarak dan 4 baris yang melanggar aturan konsistensi jarak dengan waktu tempuh. Selain itu, terdapat 23 baris yang tidak sesuai dengan kategori transportasi yang ditetapkan. Pelanggaran tersebut belum tentu merupakan kesalahan data, karena bisa terjadi akibat perbedaan penulisan kategori atau nilai yang tidak biasa. Oleh karena itu, saya perlu memeriksa kategori transportasi yang sebenarnya serta meninjau kembali data jarak dan waktu tempuh. Pemeriksaan dilakukan menggunakan indeks baris tanpa menampilkan nama responden untuk menjaga privasi.

## Bagian 5 — Kesenjangan Data dan Etika

```python
# TODO: untuk minimal 1 pertanyaan Anda yang TIDAK dapat dijawab sepenuhnya,
# tunjukkan dengan kode kolom apa yang dibutuhkan tetapi tidak ada di df.columns

# Bagian 5: Memeriksa kesenjangan data

kolom_dibutuhkan = [
    "kondisi_lalu_lintas",
    "kapasitas_parkir"
]

for kolom in kolom_dibutuhkan:
    if kolom in df.columns:
        print(f"{kolom}: Tersedia")
    else:
        print(f"{kolom}: Tidak tersedia")

```

| Pertanyaan                                               | Data yang dibutuhkan                              | Tersedia?      | Tindakan yang bertanggung jawab                                                             |
| -------------------------------------------------------- | ------------------------------------------------- | -------------- | ------------------------------------------------------------------------------------------- |
| Apakah kemacetan memengaruhi waktu perjalanan mahasiswa? | Kondisi lalu lintas dan waktu keberangkatan       | Tidak tersedia | Tidak menyimpulkan pengaruh kemacetan. Kumpulkan data tambahan jika diperlukan.             |
| Apakah kapasitas parkir kampus mencukupi?                | Kapasitas parkir dan jumlah kendaraan yang parkir | Tidak tersedia | Tidak menyimpulkan kapasitas parkir hanya dari moda transportasi. Lakukan pendataan parkir. |


```python
# (Opsional) Uji kuasi-identitas: berapa banyak kombinasi yang hanya berisi 1 orang?
# Petunjuk: df.groupby([...]).size(), lalu hitung kelompok berukuran 1

# Uji kuasi-identitas
kolom_identitas = [
    "usia",
    "semester",
    "moda_transportasi",
    "jarak_km",
    "waktu_tempuh_menit"
]

ukuran_kelompok = (
    df.groupby(kolom_identitas, dropna=False)
      .size()
)

jumlah_kombinasi_unik = (ukuran_kelompok == 1).sum()

print("Jumlah kombinasi yang hanya dimiliki satu baris:",
      jumlah_kombinasi_unik)

print("Total kombinasi yang terbentuk:",
      len(ukuran_kelompok))

```
